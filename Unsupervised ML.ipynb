1)   K-Means Clustering
K-Means is an Unsupervised Machine Learning Algorithm used to divide similar data into different groups called clusters.

K = Number of clusters (groups)
Means = Center point of each cluster (Centroid)

Why We Use K-Means?
When we have a large amount of data and don't know the categories, K-Means automatically groups similar data points together.

Example Students' Marks:
20, 25, 30, 75, 80, 85

K-Means can divide them into:
Cluster 1 = 20, 25, 30
Cluster 2 = 75, 80, 85

Step 1: Choose K
how many groups you want.
K = 2 means 2 clusters.

Step 2: Select Centroids
Choose random center points.

Centroid 1 = 25
Centroid 2 = 80

Step 3: Calculate Distance
Find the distance between each data point and the centroids.
Formula
This is called Euclidean Distance.


Flow Diagram
Dataset
   ↓
Choose K Value
   ↓
Select Random Centroids
   ↓
Calculate Distance
   ↓
Assign Points to Nearest Centroid
   ↓
Calculate New Centroids
   ↓
Repeat Until Clusters Stop Changing
   ↓
Final Clusters
Advantages
Simple and easy to understand
Fast algorithm
Works well on large datasets
Disadvantages
Need to choose K value before training
Sensitive to outliers
Different centroids can give different results

2) Hierarchical Clustering same but It creates a tree-like structure called a dendrogram to visualize the clustering process.
   Uses distance between clusters
   we do not need to specify K (number of clusters) at the beginning.

   Types of Hierarchical Clustering
1. Agglomerative Clustering (Bottom-Up)
Each data point starts as its own cluster.
Merge the closest clusters repeatedly.
Continue until all points become one cluster.
Single Points
     ↓
Merge Closest Clusters
     ↓
Larger Clusters
     ↓
One Final Cluster
2. Divisive Clustering (Top-Down)
Start with one large cluster.
Split into smaller clusters.
Continue until each point becomes a separate cluster.
One Big Cluster
      ↓
Split Clusters
      ↓
Smaller Clusters
      ↓
Individual Points
Working Example
Dataset:
20, 25, 30, 75, 80, 85
Step 1
Each point becomes a cluster.
{20} {25} {30} {75} {80} {85}
Step 2
Find nearest clusters.
20 and 25
75 and 80
Step 3

Merge them.

{20,25} {30} {75,80} {85}
Step 4

Merge again.

{20,25,30} {75,80,85}
Step 5

Continue until one cluster remains.

{20,25,30,75,80,85}
Distance Formula
Euclidean Distance
Used to find which clusters are closest.

Dendrogram

A graphical tree showing how clusters merge.

           All Data
          /        \
   {20,25,30}   {75,80,85}
      /   \         /   \
 {20,25} {30} {75,80} {85}
   /  \           /  \
 20  25         75  80
Flow
Dataset
   ↓
Create Individual Clusters
   ↓
Calculate Distance
   ↓
Find Closest Clusters
   ↓
Merge Clusters
   ↓
Repeat Process
   ↓
Create Dendrogram
   ↓
Final Clusters
Advantages
No need to choose K initially
Easy visualization using dendrogram
Good for small datasets
Shows relationship between clusters 

3) DBSCAN is an Unsupervised Machine Learning Clustering Algorithm that groups data points based on density (how closely packed points are).

Does NOT require number of clusters (K)
Can find irregular shaped clusters
Can detect outliers (noise points)

Instead of centroids (like K-Means), DBSCAN forms clusters based on:

“If points are close together → they form a cluster”
“If a point is isolated → it becomes noise”
How DBSCAN works
Step 1: Pick a random point
Step 2: Draw (ε)Epsilon radius around it
Step 3: Check neighbors
If neighbors ≥ MinPts → Core point → Start cluster
If not → mark as noise or border
Step 4: Expand cluster
Add all density-connected points
Repeat for new core points
Step 5: Repeat for all points

4) NLP Complete Notes (OneNote Style)
NLP Definition
NLP (Natural Language Processing) is a branch of AI that helps computers understand, analyze, and process human language.

Examples:
• Google Translate
• Siri
• Alexa
• Grammarly
this are the NLP pipeline 

1. Tokenization
Definition:
Breaking a sentence into smaller pieces called tokens.
Example:
Sentence: I am learning NLP today

After Tokenization:
['I', 'am', 'learning', 'NLP', 'today']
Types:
1. Word Tokenization
I love Python → ['I','love','Python']
2. Sentence Tokenization
I love Python. It is easy.
→ ['I love Python.','It is easy.']

2. Stop Words Removal
Definition:
Stop words are common words that do not add much meaning.
Examples:
is, am, are, the, a, an, of, for, to, in
Example:
Original: I am learning NLP in Python
After removing stop words:
learning NLP Python

Advantages:
• Reduces unnecessary words
• Makes model faster
• Improves accuracy

3. POS Tagging (Parts of Speech)
Definition:
Identifying the grammatical role of each word.
Example:
Sandeep is learning Python

Word      POS
Sandeep   Noun
is        Verb
learning  Verb
Python    Noun

Common POS Tags:

Noun (NN):
Dog, School, India

Verb (VB):
Run, Eat, Play

Adjective (JJ):
Beautiful, Smart, Tall

Adverb (RB):
Quickly, Slowly

Pronoun (PRP):
He, She, I, They

4. Chunking
Definition:
Grouping related words together after POS tagging.
Also called Shallow Parsing.

Example:
The smart student learns Python

POS Tags:
The      DT  Words that come before nouns.
smart    JJ  Describes a noun.
student  NN  Person, Place, Thing, Animal.
learns   VB  Action word.
Python   NN  Person, Place, Thing, Animal.

Chunk:
[The smart student]
→ Noun Phrase (NP)

Purpose:
• Finds meaningful groups
• Creates Noun Phrases and Verb Phrases


5. Stemming
Definition:
Reducing a word to its root form.
Removes prefixes and suffixes.

Examples:
Playing → Play
Played → Play
Player → Play
Studies → Studi

Note:
'Studi' is not a valid English word.
This is the limitation of stemming.

Advantages:
• Fast
• Simple

Disadvantages:
• Can create invalid words


6. Porter Stemmer
Most popular stemmer.

Example:
Running → Run

Advantages:
• Fast
• Simple
• Commonly used


7. Lancaster Stemmer
More aggressive stemmer.

Example:
Maximum → Maxim

Advantages:
• Very fast

Disadvantages:
• May remove too many characters
• Can produce incorrect roots


8. Snowball Stemmer
Improved version of Porter Stemmer.

Example:
Running → Run

Advantages:
• Better accuracy
• Supports multiple languages


9. Lemmatization
Definition:
Converts words into their actual dictionary form (Lemma).

Examples:
Running → Run
Better → Good
Studies → Study
Went → Go

Example Sentence:
Children are running

After Lemmatization:
Child be run

Advantages:
• Produces meaningful words
• More accurate than stemming


10. Stemming vs Lemmatization
Word: Studies

Stemming:
Studies → Studi

Lemmatization:
Studies → Study

Comparison:

Stemming:
• Fast
• Rule Based
• Less Accurate

Lemmatization:
• Slower
• Dictionary Based
• More Accurate
